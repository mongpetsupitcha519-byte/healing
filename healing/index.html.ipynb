```python
import React, { useState, useEffect, useRef } from 'react';
import { Moon, Sun, Sparkles, MessageCircle, Heart, Send, User, Bot, Layers, Wind, Feather, Globe, ChevronLeft, Briefcase, Flower2 } from 'lucide-react';

const apiKey = ""; // API Key จะถูกใส่โดยระบบอัตโนมัติ

// --- Translations ---
const translations = {
  th: {
    appSubtitle: "พื้นที่เร้นลับและการเยียวยา",
    tabQuotes: "คำคม",
    tabTherapist: "ที่ปรึกษา",
    tabDivination: "พยากรณ์",
    tabSelfLove: "ส่งความรู้สึก",
    btnWind: "ปล่อยให้สายลมพัดพาคำคมใหม่มา",
    therapistSelect: "เลือกผู้เชี่ยวชาญที่คุณอยากพูดคุยด้วย",
    therapistOnline: "กำลังออนไลน์",
    inputPlaceholder: "พิมพ์ข้อความของคุณที่นี่...",
    divinationTitle: "พยากรณ์นำทางชีวิต",
    divinationSub: "ค้นหาคำตอบผ่านหน้าไพ่",
    divinationStep1: "จดจ่อกับสิ่งที่คุณอยากรู้ แล้วพิมพ์คำถามลงไป",
    divinationInput: "เช่น ตอนนี้ฉันควรโฟกัสเรื่องอะไร? หรือ ความรักจะเป็นอย่างไร?",
    divinationDeckTitle: "เลือกศาสตร์แห่งการพยากรณ์",
    divinationTarot: "ไพ่ทาโรต์ (Tarot)",
    divinationPoker: "ไพ่ป๊อก (Poker)",
    divinationDrawing: "กำลังเชื่อมต่อกับจักรวาล...",
    divinationReading: "กำลังอ่านหน้าไพ่...",
    divinationAskAgain: "ถามคำถามใหม่",
    selfLoveTitle: "จดหมายถึงวิญญาณ",
    selfLoveSub: "พื้นที่เขียนระบายความรู้สึก ส่งถึงคนที่คุณรัก หรือกอดตัวเองผ่านตัวอักษร",
    selfLoveInput: "ความรู้สึกที่เอ่อล้นอยู่ในใจตอนนี้คือ...",
    selfLoveSend: "ปล่อยความรู้สึกนี้ออกไป",
    selfLoveLoading: "กำลังซึมซับทุกตัวอักษรของคุณ...",
    selfLoveNew: "เขียนจดหมายฉบับใหม่",
    errAPI: "ขออภัยค่ะ ระบบกำลังพักผ่อน ลองใหม่อีกครั้งนะคะ"
  },
  en: {
    appSubtitle: "Mystic & Healing Space",
    tabQuotes: "Quotes",
    tabTherapist: "Counselors",
    tabDivination: "Divination",
    tabSelfLove: "Self-Love",
    btnWind: "Let the wind bring a new quote",
    therapistSelect: "Select an expert you'd like to speak with",
    therapistOnline: "IS ONLINE",
    inputPlaceholder: "Type your message here...",
    divinationTitle: "The Oracle",
    divinationSub: "Seek clarity through the cards",
    divinationStep1: "Focus on what you wish to know, then type your question.",
    divinationInput: "e.g., What should I focus on right now? or How is my love life?",
    divinationDeckTitle: "Choose your deck",
    divinationTarot: "Tarot Deck",
    divinationPoker: "Poker Deck",
    divinationDrawing: "Connecting with the universe...",
    divinationReading: "Interpreting the stars...",
    divinationAskAgain: "Ask another question",
    selfLoveTitle: "Letters to the Soul",
    selfLoveSub: "A space to express feelings, write to a loved one, or embrace yourself.",
    selfLoveInput: "The feelings overflowing in my heart right now are...",
    selfLoveSend: "Release these feelings",
    selfLoveLoading: "Absorbing your every word...",
    selfLoveNew: "Write a new letter",
    errAPI: "Sorry, the system is resting. Please try again later."
  }
};

// --- Data Constants ---
const QUOTES = [
  { th: "ไม่เป็นไรเลยที่วันนี้จะไม่ได้ทำอะไรเป็นชิ้นเป็นอัน การพักผ่อนก็คือความสำเร็จอย่างหนึ่ง", en: "It is completely okay if you didn't achieve anything today. Resting is also an achievement." },
  { th: "ท้องฟ้ายังมีวันมืดมิดและสว่างไสว จิตใจของคุณก็เช่นกัน", en: "The sky has its dark and bright days, and so does your mind." },
  { th: "การสูญเสียใครสักคนไป ไม่ได้แปลว่าคุณสูญเสียคุณค่าในตัวเอง", en: "Losing someone does not mean you have lost your own value." },
  { th: "เขาแค่เดินออกจากชีวิตคุณ ไม่ได้เดินเอาอนาคตของคุณไปด้วย", en: "They merely walked out of your life; they didn't walk away with your future." },
  { th: "ล้มเจ็ดครั้ง ลุกแปดครั้ง นั่นแหละคือวิถีของผู้ชนะ", en: "Fall seven times, stand up eight. That is the way of a winner." },
  { th: "ดอกไม้ยังใช้เวลาในการผลิบาน ความสำเร็จของคุณก็เช่นกัน", en: "Flowers take time to bloom, and so does your success." },
  { th: "คุณคือโปรเจกต์ที่สำคัญที่สุดที่คุณจะได้ลงมือทำ", en: "You are the most important project you will ever work on." },
  { th: "ความสวยงามที่แท้จริง คือการที่คุณภูมิใจในความเป็นตัวเอง", en: "True beauty is being proud of who you are." },
  { th: "สายน้ำไม่ไหลย้อนกลับ เวลาและคำพูดก็เช่นกัน ทำวันนี้ให้ดีที่สุด", en: "A river doesn't flow backward; neither do time and words. Make today your best." },
  { th: "ความเงียบในบางครั้ง คือคำตอบที่ดังที่สุด", en: "Silence, sometimes, is the loudest answer." }
  // (สามารถเพิ่มคำคมให้ครบ 500 ได้ในโครงสร้างนี้)
];

const THERAPISTS = [
  { 
    id: 'luna', icon: Heart, 
    name: { th: 'ลูน่า (Luna)', en: 'Luna' }, 
    role: { th: 'ผู้รับฟังด้วยหัวใจ (ทั่วไป)', en: 'Empathetic Listener (General)' },
    desc: { th: 'พร้อมรับฟังทุกเรื่องราว ความเหนื่อยล้า โดยไม่ตัดสินคุณ', en: 'Ready to listen to your exhaustion and stories without judgment.' },
    prompt: 'You are Luna, an incredibly gentle and empathetic listener. You do not judge. You provide warmth and comfort.'
  },
  { 
    id: 'dr_anya', icon: Flower2, 
    name: { th: 'ดร. อัญญา (Dr. Anya)', en: 'Dr. Anya' }, 
    role: { th: 'ผู้เชี่ยวชาญด้านความรัก (Love)', en: 'Love & Relationship Expert' },
    desc: { th: 'ช่วยเยียวยาบาดแผลทางใจ อาการอกหัก และปัญหาความสัมพันธ์', en: 'Helps heal emotional wounds, heartbreak, and relationship issues.' },
    prompt: 'You are Dr. Anya, an expert relationship and love psychologist. You give actionable, comforting advice for heartbreak and relationships.'
  },
  { 
    id: 'coach_v', icon: Briefcase, 
    name: { th: 'โค้ช วี (Coach V)', en: 'Coach V' }, 
    role: { th: 'ที่ปรึกษาด้านการงาน & แรงบันดาลใจ', en: 'Career & Motivation Coach' },
    desc: { th: 'ปลุกไฟในวันที่ท้อแท้ หมดไฟ หรือสับสนในเส้นทางชีวิต', en: 'Reignites your fire on days of burnout, discouragement, or career confusion.' },
    prompt: 'You are Coach V, an empowering and motivational career coach. You help people overcome burnout, find purpose, and build resilience.'
  },
  { 
    id: 'master_zen', icon: Wind, 
    name: { th: 'มาสเตอร์ เซน (Master Zen)', en: 'Master Zen' }, 
    role: { th: 'ผู้ชี้แนะสัจธรรม & การปล่อยวาง', en: 'Spiritual & Letting Go Guide' },
    desc: { th: 'ให้มุมมองทางปรัชญา ช่วยให้จิตใจสงบและเรียนรู้ที่จะปล่อยวาง', en: 'Provides philosophical perspectives to help calm the mind and learn to let go.' },
    prompt: 'You are Master Zen, a wise, philosophical guide. You speak in profound metaphors, teaching acceptance, mindfulness, and the art of letting go.'
  }
];

const TAROT_DECK = ["The Fool", "The Magician", "The High Priestess", "The Empress", "The Emperor", "The Hierophant", "The Lovers", "The Chariot", "Strength", "The Hermit", "Wheel of Fortune", "Justice", "The Hanged Man", "Death", "Temperance", "The Devil", "The Tower", "The Star", "The Moon", "The Sun", "Judgement", "The World"];
const POKER_DECK = ["Hearts ♥️", "Diamonds ♦️", "Clubs ♣️", "Spades ♠️"].flatMap(suit => ["Ace", "2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King"].map(val => `${val} of ${suit}`));


// --- Main Application ---
export default function App() {
  const [isDarkMode, setIsDarkMode] = useState(true);
  const [activeTab, setActiveTab] = useState('quotes');
  const [lang, setLang] = useState('th');
  const tStr = translations[lang];

  // Colors requested by User (Sapphire/Royal Blue vs Burgundy/Champagne Beige)
  const themeStyles = isDarkMode ? {
    bg: '#112250', // Royal Blue
    surface: '#3C507D', // Sapphire
    card: 'rgba(60, 80, 125, 0.4)', // Glassmorphism Sapphire
    accent: '#E0C58F', // Quicksand (Gold)
    textMain: '#F5F0E9', // Swan Wing
    textMuted: '#D9CBC2', // Shellstone
    border: 'rgba(224, 197, 143, 0.25)', // Gold border
    glow: 'rgba(224, 197, 143, 0.1)',
  } : {
    bg: '#F2E5C6', // Champagne Beige
    surface: '#F2D9A0', // Sand Gold
    card: 'rgba(242, 217, 160, 0.5)', // Glassmorphism Sand
    accent: '#75162D', // Burgundy
    textMain: '#3B010B', // Deep Wine
    textMuted: '#560B18', // Dark Maroon
    border: 'rgba(117, 22, 45, 0.25)', // Burgundy border
    glow: 'rgba(117, 22, 45, 0.08)',
  };

  // API Helper
  const callGeminiAPI = async (prompt, systemInstruction) => {
    const payload = {
      contents: [{ parts: [{ text: prompt }] }],
      systemInstruction: { parts: [{ text: `${systemInstruction}\n\nIMPORTANT: You must respond entirely in the language corresponding to this code: ${lang} (th = Thai, en = English).` }] }
    };
    try {
      const response = await fetch(`https://generativelanguage.googleapis.com/v1beta/models/gemini-2.5-flash-preview-09-2025:generateContent?key=${apiKey}`, {
        method: 'POST', headers: { 'Content-Type': 'application/json' }, body: JSON.stringify(payload)
      });
      if (!response.ok) throw new Error();
      const data = await response.json();
      return data.candidates?.[0]?.content?.parts?.[0]?.text || tStr.errAPI;
    } catch (error) {
      return tStr.errAPI;
    }
  };

  return (
    <>
      <style>{`
        @import url('https://fonts.googleapis.com/css2?family=Playfair+Display:ital,wght@0,400;0,600;1,400&family=Prompt:wght@300;400;500&display=swap');
        .font-serif { font-family: 'Playfair Display', serif; }
        .font-sans { font-family: 'Prompt', sans-serif; }
        .hide-scrollbar::-webkit-scrollbar { display: none; }
        .hide-scrollbar { -ms-overflow-style: none; scrollbar-width: none; }
        
        .glass-panel {
          backdrop-filter: blur(20px);
          -webkit-backdrop-filter: blur(20px);
        }

        .fade-in { animation: fadeIn 0.6s cubic-bezier(0.16, 1, 0.3, 1) forwards; }
        @keyframes fadeIn {
          from { opacity: 0; transform: translateY(15px); }
          to { opacity: 1; transform: translateY(0); }
        }

        /* Mystic Background Gradient */
        .ambient-bg {
          position: fixed;
          top: 0; left: 0; width: 100vw; height: 100vh;
          z-index: 0;
          background: radial-gradient(circle at 20% 30%, ${themeStyles.glow}, transparent 60%),
                      radial-gradient(circle at 80% 70%, ${themeStyles.glow}, transparent 60%);
          transition: background 1s ease;
        }
      `}</style>

      <div className="min-h-screen transition-colors duration-700 font-sans flex flex-col relative"
           style={{ backgroundColor: themeStyles.bg, color: themeStyles.textMain }}>
        
        <div className="ambient-bg pointer-events-none"></div>

        {/* Elegant Header */}
        <header className="px-6 py-6 flex justify-between items-center z-10 sticky top-0 glass-panel"
                style={{ borderBottom: `1px solid ${themeStyles.border}` }}>
          
          {/* Logo Section */}
          <div className="flex items-center gap-3 cursor-pointer group" onClick={() => setActiveTab('quotes')}>
             <div className="relative w-10 h-10 flex items-center justify-center">
                <Sun className="absolute transition-transform duration-1000 group-hover:rotate-180" size={28} style={{ color: themeStyles.accent, opacity: 0.8 }} />
                <Moon className="absolute transition-transform duration-1000 -rotate-45 group-hover:rotate-90" size={16} style={{ color: themeStyles.textMain }} />
             </div>
             <div className="flex flex-col">
              <h1 className="text-2xl font-serif tracking-[0.15em] font-bold" style={{ color: themeStyles.accent }}>
                XAVIROBZONE
              </h1>
              <p className="text-[10px] tracking-[0.2em] uppercase opacity-70 font-light mt-0.5">
                {tStr.appSubtitle}
              </p>
            </div>
          </div>
          
          {/* Controls */}
          <div className="flex items-center gap-3">
             <button 
                onClick={() => setLang(lang === 'th' ? 'en' : 'th')}
                className="flex items-center gap-1 px-3 py-1.5 rounded-full border text-xs font-medium uppercase tracking-wider transition-all hover:bg-opacity-20"
                style={{ borderColor: themeStyles.border, color: themeStyles.textMain, backgroundColor: themeStyles.surface }}
              >
                <Globe size={14} /> {lang}
              </button>
            <button 
              onClick={() => setIsDarkMode(!isDarkMode)}
              className="w-10 h-10 rounded-full flex items-center justify-center transition-all duration-700 hover:rotate-90 border"
              style={{ backgroundColor: themeStyles.card, color: themeStyles.accent, borderColor: themeStyles.border }}
            >
              {isDarkMode ? <Sun size={18} /> : <Moon size={18} />}
            </button>
          </div>
        </header>

        {/* Navigation */}
        <div className="z-10 py-6 px-4 flex justify-center w-full">
          <div className="flex overflow-x-auto gap-2 p-1.5 rounded-full hide-scrollbar glass-panel"
               style={{ backgroundColor: themeStyles.card, border: `1px solid ${themeStyles.border}` }}>
            {[
              { id: 'quotes', icon: Feather, label: tStr.tabQuotes },
              { id: 'therapist', icon: MessageCircle, label: tStr.tabTherapist },
              { id: 'divination', icon: Layers, label: tStr.tabDivination },
              { id: 'selflove', icon: Heart, label: tStr.tabSelfLove }
            ].map((tab) => (
              <button
                key={tab.id}
                onClick={() => setActiveTab(tab.id)}
                className={`flex items-center gap-2 px-6 py-2.5 rounded-full text-sm font-medium transition-all duration-500`}
                style={{ 
                  backgroundColor: activeTab === tab.id ? themeStyles.accent : 'transparent',
                  color: activeTab === tab.id ? themeStyles.bg : themeStyles.textMain,
                  boxShadow: activeTab === tab.id ? `0 4px 15px ${themeStyles.glow}` : 'none'
                }}
              >
                <tab.icon size={16} strokeWidth={activeTab === tab.id ? 2.5 : 1.5} />
                <span className={activeTab === tab.id ? 'font-semibold' : ''}>{tab.label}</span>
              </button>
            ))}
          </div>
        </div>

        {/* Main Content Area */}
        <main className="flex-1 p-4 md:p-8 z-10 flex flex-col max-w-5xl w-full mx-auto pb-24 relative">
          {activeTab === 'quotes' && <QuotesTab themeStyles={themeStyles} lang={lang} tStr={tStr} />}
          {activeTab === 'therapist' && <TherapistTab themeStyles={themeStyles} lang={lang} tStr={tStr} callGeminiAPI={callGeminiAPI} />}
          {activeTab === 'divination' && <DivinationTab themeStyles={themeStyles} lang={lang} tStr={tStr} callGeminiAPI={callGeminiAPI} />}
          {activeTab === 'selflove' && <SelfLoveTab themeStyles={themeStyles} lang={lang} tStr={tStr} callGeminiAPI={callGeminiAPI} />}
        </main>
      </div>
    </>
  );
}

// ==========================================
// 1. QUOTES TAB (Human-Designed Minimalist Gallery)
// ==========================================
function QuotesTab({ themeStyles, lang, tStr }) {
  const [currentQuote, setCurrentQuote] = useState(QUOTES[0]);
  const [isFading, setIsFading] = useState(false);

  const getRandomQuote = () => {
    setIsFading(true);
    setTimeout(() => {
      const randomIndex = Math.floor(Math.random() * QUOTES.length);
      setCurrentQuote(QUOTES[randomIndex]);
      setIsFading(false);
    }, 400);
  };

  useEffect(() => { getRandomQuote(); }, []);

  return (
    <div className="flex-1 flex flex-col items-center justify-center fade-in">
      <div 
        className="glass-panel rounded-3xl p-10 md:p-16 max-w-3xl w-full relative overflow-hidden"
        style={{ backgroundColor: themeStyles.card, border: `1px solid ${themeStyles.border}`, boxShadow: `0 20px 40px -10px ${themeStyles.glow}` }}
      >
        <div className="absolute top-4 left-6 text-[8rem] font-serif opacity-10 leading-none pointer-events-none" style={{ color: themeStyles.accent }}>"</div>

        <div className={`transition-opacity duration-500 ease-in-out relative z-10 ${isFading ? 'opacity-0' : 'opacity-100'}`}>
          <p className="text-2xl md:text-3xl font-light mb-8 leading-relaxed text-center" style={{ color: themeStyles.textMain }}>
            "{currentQuote[lang]}"
          </p>
          <div className="flex items-center justify-center gap-4">
            <div className="h-[1px] w-12" style={{ backgroundColor: themeStyles.accent }}></div>
             <p className="text-sm font-serif italic text-center opacity-70" style={{ color: themeStyles.accent }}>
              Xavirobzone Collection
            </p>
            <div className="h-[1px] w-12" style={{ backgroundColor: themeStyles.accent }}></div>
          </div>
        </div>
      </div>

      <button 
        onClick={getRandomQuote}
        className="mt-12 flex items-center justify-center gap-3 px-8 py-4 rounded-full text-sm font-medium transition-all duration-300 hover:scale-105 border"
        style={{ backgroundColor: themeStyles.surface, color: themeStyles.textMain, borderColor: themeStyles.border }}
      >
        <Wind size={18} className={isFading ? 'animate-spin' : ''} style={{ color: themeStyles.accent }} />
        {tStr.btnWind}
      </button>
    </div>
  );
}

// ==========================================
// 2. THERAPIST TAB (Multi-Persona)
// ==========================================
function TherapistTab({ themeStyles, lang, tStr, callGeminiAPI }) {
  const [selectedTherapist, setSelectedTherapist] = useState(null);
  const [messages, setMessages] = useState([]);
  const [input, setInput] = useState("");
  const [isLoading, setIsLoading] = useState(false);
  const messagesEndRef = useRef(null);

  const scrollToBottom = () => messagesEndRef.current?.scrollIntoView({ behavior: "smooth" });
  useEffect(() => { scrollToBottom(); }, [messages]);

  const selectTherapist = (therapist) => {
    setSelectedTherapist(therapist);
    const greeting = lang === 'th' ? `สวัสดีค่ะ ฉันคือ ${therapist.name.th} วันนี้มีอะไรอยากให้ฉันช่วยดูแล หรืออยากเล่าให้ฟังไหมคะ?` : `Hello, I'm ${therapist.name.en}. How can I support you today?`;
    setMessages([{ role: 'ai', text: greeting }]);
  };

  const handleSend = async () => {
    if (!input.trim() || isLoading) return;
    const userText = input.trim();
    setInput("");
    setMessages(prev => [...prev, { role: 'user', text: userText }]);
    setIsLoading(true);

    const history = messages.slice(-5).map(m => `${m.role === 'ai' ? 'Therapist' : 'User'}: ${m.text}`).join('\n');
    const response = await callGeminiAPI(`${history}\nUser: ${userText}\nTherapist:`, selectedTherapist.prompt);
    
    setMessages(prev => [...prev, { role: 'ai', text: response }]);
    setIsLoading(false);
  };

  if (!selectedTherapist) {
    return (
      <div className="flex-1 flex flex-col items-center justify-center fade-in w-full">
        <h2 className="text-2xl font-serif mb-8 text-center" style={{ color: themeStyles.accent }}>{tStr.therapistSelect}</h2>
        <div className="grid grid-cols-1 md:grid-cols-2 gap-6 w-full max-w-4xl">
          {THERAPISTS.map((t) => (
            <button 
              key={t.id} 
              onClick={() => selectTherapist(t)}
              className="flex flex-col items-center text-center p-8 rounded-3xl glass-panel transition-all hover:scale-[1.02] border"
              style={{ backgroundColor: themeStyles.card, borderColor: themeStyles.border }}
            >
              <t.icon size={40} className="mb-4" style={{ color: themeStyles.accent }} />
              <h3 className="text-xl font-bold mb-2">{t.name[lang]}</h3>
              <p className="text-sm font-semibold mb-3 opacity-90" style={{ color: themeStyles.accent }}>{t.role[lang]}</p>
              <p className="text-sm opacity-70 font-light">{t.desc[lang]}</p>
            </button>
          ))}
        </div>
      </div>
    );
  }

  return (
    <div className="flex-1 flex flex-col h-[75vh] rounded-3xl overflow-hidden glass-panel border fade-in" 
         style={{ backgroundColor: themeStyles.card, borderColor: themeStyles.border, boxShadow: `0 20px 40px -10px ${themeStyles.glow}` }}>
      
      {/* Chat Header */}
      <div className="py-4 px-6 border-b flex justify-between items-center" style={{ borderColor: themeStyles.border, backgroundColor: themeStyles.surface }}>
        <button onClick={() => setSelectedTherapist(null)} className="opacity-70 hover:opacity-100 transition-opacity flex items-center gap-1 text-sm">
          <ChevronLeft size={16} /> Back
        </button>
        <div className="flex items-center gap-3">
          <div className="w-2 h-2 rounded-full animate-pulse" style={{ backgroundColor: themeStyles.accent }}></div>
          <h3 className="font-medium text-sm tracking-wide uppercase opacity-80">{selectedTherapist.name[lang]} {tStr.therapistOnline}</h3>
        </div>
        <div className="w-10"></div> {/* Spacer for centering */}
      </div>

      {/* Chat Messages */}
      <div className="flex-1 overflow-y-auto p-6 md:p-8 space-y-6">
        {messages.map((msg, idx) => (
          <div key={idx} className={`flex ${msg.role === 'user' ? 'justify-end' : 'justify-start'}`}>
            <div className={`max-w-[85%] flex flex-col ${msg.role === 'user' ? 'items-end' : 'items-start'}`}>
              <div 
                className={`p-5 rounded-2xl whitespace-pre-wrap leading-relaxed shadow-sm text-[15px]`}
                style={{ 
                  backgroundColor: msg.role === 'user' ? themeStyles.accent : themeStyles.surface, 
                  color: msg.role === 'user' ? themeStyles.bg : themeStyles.textMain,
                  borderTopRightRadius: msg.role === 'user' ? '4px' : '20px',
                  borderTopLeftRadius: msg.role === 'ai' ? '4px' : '20px',
                }}
              >
                {msg.text}
              </div>
            </div>
          </div>
        ))}
        {isLoading && (
          <div className="flex justify-start fade-in">
             <div className="p-5 rounded-2xl rounded-tl-sm flex gap-2 items-center" style={{ backgroundColor: themeStyles.surface }}>
                <span className="w-2 h-2 rounded-full animate-bounce" style={{ backgroundColor: themeStyles.accent }}></span>
                <span className="w-2 h-2 rounded-full animate-bounce delay-100" style={{ backgroundColor: themeStyles.accent }}></span>
                <span className="w-2 h-2 rounded-full animate-bounce delay-200" style={{ backgroundColor: themeStyles.accent }}></span>
             </div>
          </div>
        )}
        <div ref={messagesEndRef} />
      </div>

      {/* Chat Input */}
      <div className="p-4 border-t" style={{ borderColor: themeStyles.border, backgroundColor: themeStyles.surface }}>
        <div className="flex gap-3 relative max-w-4xl mx-auto">
          <input 
            type="text" value={input} onChange={(e) => setInput(e.target.value)} onKeyPress={(e) => e.key === 'Enter' && handleSend()}
            placeholder={tStr.inputPlaceholder}
            className="flex-1 p-4 px-6 rounded-full outline-none text-sm transition-shadow focus:shadow-inner"
            style={{ backgroundColor: themeStyles.bg, color: themeStyles.textMain, border: `1px solid ${themeStyles.border}` }}
          />
          <button 
            onClick={handleSend} disabled={isLoading || !input.trim()}
            className="w-14 rounded-full flex items-center justify-center transition-transform hover:scale-105 disabled:opacity-30"
            style={{ backgroundColor: themeStyles.accent, color: themeStyles.bg }}
          >
            <Send size={18} />
          </button>
        </div>
      </div>
    </div>
  );
}

// ==========================================
// 3. DIVINATION TAB (Restored Cards & Clear Selection)
// ==========================================
function DivinationTab({ themeStyles, lang, tStr, callGeminiAPI }) {
  const [step, setStep] = useState(1); 
  const [question, setQuestion] = useState("");
  const [deckType, setDeckType] = useState(""); 
  const [drawnCards, setDrawnCards] = useState([]);
  const [reading, setReading] = useState("");
  const [isReading, setIsReading] = useState(false);

  const drawCards = async (type) => {
    setDeckType(type);
    setStep(3); // Animate drawing
    let deck = type === 'tarot' ? [...TAROT_DECK] : [...POKER_DECK];
    let selected = [];
    for(let i=0; i<3; i++) {
      const idx = Math.floor(Math.random() * deck.length);
      selected.push(deck[idx]);
      deck.splice(idx, 1);
    }
    setTimeout(() => {
      setDrawnCards(selected);
      getAIReading(selected, type);
    }, 2500); 
  };

  const getAIReading = async (cards, type) => {
    setStep(4);
    setIsReading(true);
    const systemPrompt = `You are a sophisticated fortune teller. The user asked: "${question}". They drew 3 ${type} cards: ${cards.join(', ')}. Provide a deeply insightful, structured reading. Use elegant language. Be uplifting.`;
    const response = await callGeminiAPI(`Interpret these cards based on the question.`, systemPrompt);
    setReading(response);
    setIsReading(false);
  };

  return (
    <div className="flex-1 flex flex-col items-center justify-start fade-in w-full">
      <div className="w-full max-w-4xl space-y-10">
        
        <div className="text-center space-y-2">
          <h2 className="text-3xl font-serif" style={{ color: themeStyles.accent }}>{tStr.divinationTitle}</h2>
          <p className="text-sm opacity-60 uppercase tracking-widest">{tStr.divinationSub}</p>
        </div>

        {/* Combined Step 1 & 2 */}
        {step === 1 && (
          <div className="glass-panel p-8 md:p-12 rounded-3xl border mx-auto max-w-2xl text-center" style={{ backgroundColor: themeStyles.card, borderColor: themeStyles.border }}>
            <p className="mb-6 font-medium text-lg opacity-90">{tStr.divinationStep1}</p>
            <textarea
              value={question} onChange={(e) => setQuestion(e.target.value)}
              placeholder={tStr.divinationInput}
              className="w-full p-6 rounded-2xl outline-none resize-none h-32 mb-8 bg-opacity-50 border transition-all focus:ring-1"
              style={{ backgroundColor: themeStyles.bg, color: themeStyles.textMain, borderColor: themeStyles.border }}
            />
            
            <div className={`transition-all duration-700 overflow-hidden ${question.trim() ? 'max-h-96 opacity-100' : 'max-h-0 opacity-0'}`}>
               <div className="w-full h-[1px] mb-8" style={{ backgroundColor: themeStyles.border }}></div>
               <p className="text-sm opacity-70 mb-4 uppercase tracking-widest">{tStr.divinationDeckTitle}</p>
               <div className="flex flex-col sm:flex-row gap-4 justify-center">
                  <button onClick={() => drawCards('tarot')} className="flex-1 py-6 px-4 rounded-2xl border transition-all hover:scale-105 group" style={{ borderColor: themeStyles.accent, backgroundColor: themeStyles.surface }}>
                    <Moon size={32} className="mx-auto mb-3 opacity-80 group-hover:opacity-100" style={{ color: themeStyles.accent }}/>
                    <span className="font-serif block font-bold text-lg">{tStr.divinationTarot}</span>
                  </button>
                  <button onClick={() => drawCards('poker')} className="flex-1 py-6 px-4 rounded-2xl border transition-all hover:scale-105 group" style={{ borderColor: themeStyles.accent, backgroundColor: themeStyles.surface }}>
                    <div className="flex justify-center gap-2 mb-3 opacity-80 group-hover:opacity-100" style={{ color: themeStyles.accent }}>
                      <span className="text-2xl">♠️</span><span className="text-2xl">♥️</span>
                    </div>
                    <span className="font-serif block font-bold text-lg">{tStr.divinationPoker}</span>
                  </button>
               </div>
            </div>
          </div>
        )}

        {/* Step 3: Drawing Cards Animation */}
        {step === 3 && (
          <div className="py-24 flex flex-col items-center justify-center space-y-12 fade-in">
            <div className="flex gap-4 md:gap-8">
              {[1, 2, 3].map((i) => (
                <div key={i} className="w-24 h-40 md:w-32 md:h-48 rounded-xl border-2 flex items-center justify-center animate-pulse"
                     style={{ backgroundColor: themeStyles.surface, borderColor: themeStyles.accent, boxShadow: `0 0 30px ${themeStyles.glow}` }}>
                  <div className="w-16 h-24 border border-dashed rounded opacity-30" style={{ borderColor: themeStyles.textMain }}></div>
                </div>
              ))}
            </div>
            <p className="font-serif italic text-lg opacity-80" style={{ color: themeStyles.accent }}>{tStr.divinationDrawing}</p>
          </div>
        )}

        {/* Step 4: Reading Result */}
        {step === 4 && (
          <div className="space-y-12 fade-in w-full">
            {/* Show Drawn Cards Beautifully */}
            <div className="flex justify-center gap-4 md:gap-8 perspective-1000">
              {drawnCards.map((card, idx) => (
                <div key={idx} className="w-28 h-44 md:w-36 md:h-56 rounded-xl p-3 flex flex-col items-center justify-center text-center border relative overflow-hidden fade-in"
                     style={{ backgroundColor: themeStyles.surface, borderColor: themeStyles.accent, animationDelay: `${idx * 0.2}s` }}>
                  <div className="absolute inset-1 border rounded opacity-20 pointer-events-none" style={{ borderColor: themeStyles.textMain }}></div>
                  <span className="font-serif font-bold text-sm md:text-lg z-10" style={{ color: themeStyles.textMain }}>{card}</span>
                </div>
              ))}
            </div>

            <div className="glass-panel p-8 md:p-12 rounded-3xl border max-w-3xl mx-auto" style={{ backgroundColor: themeStyles.card, borderColor: themeStyles.border }}>
              {isReading ? (
                <div className="py-12 text-center flex flex-col items-center">
                  <div className="w-8 h-8 border-2 border-t-transparent rounded-full animate-spin mb-6" style={{ borderColor: themeStyles.accent, borderTopColor: 'transparent' }}></div>
                  <p className="font-serif italic text-lg opacity-80">{tStr.divinationReading}</p>
                </div>
              ) : (
                <div className="whitespace-pre-wrap leading-relaxed text-base md:text-lg font-light opacity-90">
                  {reading}
                </div>
              )}
            </div>

            {!isReading && (
              <div className="text-center pb-10">
                <button onClick={() => { setStep(1); setQuestion(""); }} className="px-6 py-2 rounded-full uppercase tracking-wider text-sm font-medium transition-all hover:bg-opacity-20 border" 
                        style={{ borderColor: themeStyles.border, color: themeStyles.textMain, backgroundColor: themeStyles.surface }}>
                  {tStr.divinationAskAgain}
                </button>
              </div>
            )}
          </div>
        )}

      </div>
    </div>
  );
}

// ==========================================
// 4. SELF LOVE TAB (Journaling)
// ==========================================
function SelfLoveTab({ themeStyles, lang, tStr, callGeminiAPI }) {
  const [message, setMessage] = useState("");
  const [response, setResponse] = useState("");
  const [isLoading, setIsLoading] = useState(false);
  const [isSent, setIsSent] = useState(false);

  const handleSend = async () => {
    if (!message.trim()) return;
    setIsLoading(true); setIsSent(true);

    const systemPrompt = `You are a deeply compassionate journal guide. The user wrote a heartfelt message. Respond by validating their feelings, praising their vulnerability, and beautifully reminding them to direct love towards themselves. Keep it structured like a letter.`;
    const aiResponse = await callGeminiAPI(`User's entry: "${message}"`, systemPrompt);
    
    setResponse(aiResponse);
    setIsLoading(false);
  };

  return (
    <div className="flex-1 flex flex-col items-center justify-center fade-in">
       <div className="w-full max-w-2xl glass-panel rounded-3xl p-8 md:p-12 border" style={{ backgroundColor: themeStyles.card, borderColor: themeStyles.border, boxShadow: `0 20px 40px -10px ${themeStyles.glow}` }}>
        
        {!isSent ? (
          <div className="space-y-8">
            <div className="text-center">
              <h2 className="text-2xl font-serif mb-3" style={{ color: themeStyles.accent }}>{tStr.selfLoveTitle}</h2>
              <p className="text-sm opacity-70 font-light">{tStr.selfLoveSub}</p>
            </div>

            <textarea
              value={message} onChange={(e) => setMessage(e.target.value)}
              placeholder={tStr.selfLoveInput}
              className="w-full p-6 rounded-2xl outline-none resize-none h-48 bg-opacity-30 border focus:bg-opacity-50 transition-all font-light text-lg leading-relaxed"
              style={{ backgroundColor: themeStyles.bg, color: themeStyles.textMain, borderColor: themeStyles.border }}
            />
            
            <button 
              onClick={handleSend} disabled={!message.trim()}
              className="w-full py-4 rounded-2xl font-bold tracking-wide transition-transform hover:scale-[1.02] disabled:opacity-30 flex justify-center items-center gap-2"
              style={{ backgroundColor: themeStyles.accent, color: themeStyles.bg }}
            >
              {tStr.selfLoveSend} <Feather size={18} />
            </button>
          </div>
        ) : (
          <div className="space-y-8 fade-in text-center">
            <Heart className="mx-auto opacity-80" size={40} style={{ color: themeStyles.accent }} strokeWidth={1.5} />
            
            <div className="text-left whitespace-pre-wrap leading-relaxed font-light text-base opacity-90 p-8 rounded-2xl border" style={{ backgroundColor: themeStyles.surface, borderColor: themeStyles.border }}>
              {isLoading ? (
                <div className="text-center opacity-70 py-8 animate-pulse font-serif italic">{tStr.selfLoveLoading}</div>
              ) : (
                response
              )}
            </div>

            {!isLoading && (
              <button 
                onClick={() => { setIsSent(false); setMessage(""); setResponse(""); }}
                className="px-6 py-2 rounded-full uppercase tracking-wider text-sm font-medium transition-all hover:bg-opacity-20 border" 
                style={{ borderColor: themeStyles.border, color: themeStyles.textMain, backgroundColor: themeStyles.surface }}
              >
                {tStr.selfLoveNew}
              </button>
            )}
          </div>
        )}
      </div>
    </div>
  );
}
```
